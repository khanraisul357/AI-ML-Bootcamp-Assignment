# stacking and spliting

```python
d = np.arange(27).reshape(3,9)
e = np.arange(28,55).reshape(3,9)
print(d)
print(e)
```

# use hstack() and vstack()

```python
# use hstack() and vstack()
h_stacked = np.hstack((d, e))
v_stacked = np.vstack((d, e))

print("Horizontal stacking (hstack):\n", h_stacked)
print("\nVertical stacking (vstack):\n", v_stacked)
```

# use hsplit() and vsplit()

```python
# use hsplit() and vsplit()
h_split = np.hsplit(d, 3)
v_split = np.vsplit(d, 3)

print("Horizontal split:")
for part in h_split:
    print(part)

print("\nVertical split:")
for part in v_split:
    print(part)
```
